# Predict weight based on height

```python
# Import libraries
import numpy as np 
import matplotlib.pyplot as plt 

# Input data 
X = np.array([147,150,153,158,163,165,168,170,173,175,178,180,183]).reshape((13,1))
Y = np.array([49,50,51,54,58,59,60,62,63,64,66,67,68]).reshape((13,1))

# Building Xbar
one = np.ones((X.shape[0], 1))
Xbar = np.concatenate((one,X), axis = 1)

# Calculating weights of the linear regression model 
A = np.dot(Xbar.T, Xbar)
b = np.dot(Xbar.T,Y)
w = np.dot(np.linalg.pinv(A),b)

# weights 
w_0, w_1 = w[0], w[1]
y1 = w_1*155 + w_0 
y2 = w_1*160 + w_0 

print(f"Input 155cm, true output 52kg, predicted output: {y1[0]:.2f} kg")
print(f"Input 160cm, true output 56kg, predicted output: {y2[0]:.2f} kg")

plt.plot(X,Y) 
plt.title("Graph of relationship between weights and heights"); 
```

## Check the result by scikit-learn 

```python
from sklearn import datasets, linear_model 

# Fit the model by Linear Regression 
regr = linear_model.LinearRegression()
regr.fit(X,Y) 

#Compare two results 
print(f"Scikit-learn result: w_1: {regr.coef_[0]}, w_0: {regr.intercept_}")
print(f" Our result: w_1: {w[1]}, w_0:{w[0]}")
print(f" The difference for w_1: {int((regr.coef_[0] / w[1]).item())}, "
      f"w_0: {int((regr.intercept_ / w[0]).item())}")
print("=> Almost similar")
```
